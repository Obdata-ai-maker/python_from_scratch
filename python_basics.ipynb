# Scope is litterally when python sees a variable it must decide where to look for it is it : local scope , then enclosed scope , then  global scope , then built in scope 
In Python scope is the point which you can access a variable . its what controls the lifetime of a variable and how it is resolved in different parts of the code . 

to determin scope **python** follows LEGB rule : 
1. **Local scope** : variables defined in functions or classes 
2. **Enclosing scope** : variables defined in enclosing or nested functions .
3. **Global scope** : variables defined at the top level of module
4. **Built-in-scope** : Reserved named for predefined functions .

**Local scope** : a variable declared inside a function or a class can only be accessed within the function/class

```python
def addition():
    test = 10

print(test) # test is not defined 
```

**Enclosing scope** : function nested inside a function can access the variables of the function it's nested in 


```python
def outer_function():
    x = 10 
    def inner():
        print('yes , you can access x from the outer function : ',x)
    inner()
outer_function()
```

```python
def outer_function():
    x = 10 
    def inner():
        y  = 10
        print('yes , you can access x from the outer function : ',x)
    inner()
    print(y)
outer_function()
# but the outer function cannot access y value 
```

**python determine if a function if it has acess based on definition not calling**

```python
def inner():
    if x :
        print('yes , you can access x from the outer function : ',x)
    else :
        print('yes , you can access x from the outer function : ',x)
def outer_function():
    x = 10 
    inner()
outer_function()
# inner has no access to outer function variabl  because inner was defined outside of outer
# python doesn't care about calling only where it was defined 
```

but you can resolve that outer can access inner function by creating nonlocal variable : 

```python
def outer_func():
    msg = 'Hello there!'
    res = ""  # Declare res in the enclosing scope

    def inner_func():
        nonlocal res  # Allow modification of an enclosing variable
        res = 'How are you?'
        print(msg)  # Accessing msg from outer_func()

    inner_func()
    print(res)  # Now res is accessible and modified

outer_func()
```

**Global scope** : variables defined outside any functions or classes , which can be accessed from anywhere in the program .

```python
my_var = 100

def show_var():
    print(my_var)

show_var() # 100
print(my_var) # 100
```

```python
# you can create global variable  inside a function
def show_var():
    global my_var
    my_var = 40

show_var() 
print(my_var) #40
```

```python
# or you can modify global variable  inside a function
my_var = 100
def show_var():
    global my_var
    my_var = 2

show_var() 
print(my_var) #2 instead of 100
```

**built-in scope** refers to all of Python's built-in functions, modules, and keywords, and are available anywhere in your program:

```python
print(str(45)) # '45' so str is not found in local scope , not in enclosed scope
# not in global scope but in built in scope it
print(isinstance(3, str)) # False
```

**use of break an continue statements**

**break** breaks out of the loop

```python
developer_names = ['Jess', 'Naomi', 'Tom']

for developer in developer_names:
    if developer == 'Naomi':
        break
    print(developer)
```

**continue** statement skips this iteration and moves on to the next iteration

```python
developer_names = ['Jess', 'Naomi', 'Tom']

for developer in developer_names:
    if developer == 'Naomi':
        continue
    print(developer)
```

# for comes with **else** too , not just if , so when for runs out without breaks it executes else block , if break is executed else is skipped

```python
words = ['sky', 'apple', 'rhythm', 'fly', 'orange']

for word in words:
    for letter in word:
        if letter.lower() in 'aeiou':
            print(f"'{word}' contains the vowel '{letter}'")
            break
    else:
        print(f"'{word}' has no vowels")
```
