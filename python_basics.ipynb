# Scope is litterally when python sees a variable it must decide where to look for it is it : local scope , then enclosed scope , then  global scope , then built in scope 
In Python scope is the point which you can access a variable . its what controls the lifetime of a variable and how it is resolved in different parts of the code . 

to determin scope **python** follows LEGB rule : 
1. **Local scope** : variables defined in functions or classes 
2. **Enclosing scope** : variables defined in enclosing or nested functions .
3. **Global scope** : variables defined at the top level of module
4. **Built-in-scope** : Reserved named for predefined functions .

**Local scope** : a variable declared inside a function or a class can only be accessed within the function/class

```python
def addition():
    test = 10

print(test) # test is not defined 
```

**Enclosing scope** : function nested inside a function can access the variables of the function it's nested in 


```python
def outer_function():
    x = 10 
    def inner():
        print('yes , you can access x from the outer function : ',x)
    inner()
outer_function()
```

```python
def outer_function():
    x = 10 
    def inner():
        y  = 10
        print('yes , you can access x from the outer function : ',x)
    inner()
    print(y)
outer_function()
# but the outer function cannot access y value 
```

**python determine if a function if it has acess based on definition not calling**

```python
def inner():
    if x :
        print('yes , you can access x from the outer function : ',x)
    else :
        print('yes , you can access x from the outer function : ',x)
def outer_function():
    x = 10 
    inner()
outer_function()
# inner has no access to outer function variabl  because inner was defined outside of outer
# python doesn't care about calling only where it was defined 
```

but you can resolve that outer can access inner function by creating nonlocal variable : 

```python
def outer_func():
    msg = 'Hello there!'
    res = ""  # Declare res in the enclosing scope

    def inner_func():
        nonlocal res  # Allow modification of an enclosing variable
        res = 'How are you?'
        print(msg)  # Accessing msg from outer_func()

    inner_func()
    print(res)  # Now res is accessible and modified

outer_func()
```

**Global scope** : variables defined outside any functions or classes , which can be accessed from anywhere in the program .

```python
my_var = 100

def show_var():
    print(my_var)

show_var() # 100
print(my_var) # 100
```

```python
# you can create global variable  inside a function
def show_var():
    global my_var
    my_var = 40

show_var() 
print(my_var) #40
```

```python
# or you can modify global variable  inside a function
my_var = 100
def show_var():
    global my_var
    my_var = 2

show_var() 
print(my_var) #2 instead of 100
```

**built-in scope** refers to all of Python's built-in functions, modules, and keywords, and are available anywhere in your program:

```python
print(str(45)) # '45' so str is not found in local scope , not in enclosed scope
# not in global scope but in built in scope it
print(isinstance(3, str)) # False
```

# **lambda function**
lambda function are an ***anonymous inline functions***, especially useful when working with **high order function** like **map , filter**

```python
# lambda function : anonymous inline function 

lambda x : x ** 2
```

*use them with high order functions (map , filter) :*

```python
lst = [1,2,3,4,5,10,20,30]

list(map(lambda x : x**2,lst)),list(filter(lambda x : x % 2 == 0,lst))
```

***best practice : don't assign a lambda function to variable instead use regular function***  :: **why it defeats the pupose of an anonymous function**

```python
func = lambda x : x**2 # not best practice XXXXXXXXXXXXXX

lst = [1,2,3,4,5,10,20,30]
list(map(func,lst))
```

```python
# but instead use a regular function 

def func (x) : return x**2

lst = [1,2,3,4,5,10,20,30]
list(map(func,lst))
```

**avoid using complicated logic inside lambda functions**

```python
# avoid using complicated logic inside lambda functions
result = (lambda x: (x**2 + 2*x - 1) if x > 0 else (x**3 - x + 4))(13)
# the (13) means that you are invoking the function directly like : f(13)
print(result)  # 14
```

```python
# use instead 

def calculate_expression(x) : 
    if x > 0:
        return (x**2 + 2*x - 1)
    else :
        return (x**3 - x + 4)

print(calculate_expression(13))
```

#### *Common mathematical operations*

##### **check if a set is a subset or superset of another set**

```python
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}

print(your_set.issubset(my_set)) 
print(my_set.issuperset(your_set)) 
```

##### **check if a 2 sets are disjoint : they don't have have any elements in common**

```python
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}

your_set.isdisjoint(my_set)
```

##### **'|' is union and it returns a new set with all the elements from both sets**

```python
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}
my_set | your_set # full outer join
```

##### **'&' is intersection and it returns a new set with all the elements in common from both sets**

```python
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}
my_set & your_set  # join
```

##### **'-' is difference and it returns a new set with all the elements in the first set that are not in the second set**

```python
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}
my_set - your_set # left outer join excluding inner join
```

##### **'^' is symmetric difference and it returns a new set with all the elements in the first set that are not in the second set and the second set that are not in the first set**

```python
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}
my_set ^ your_set # full outer join excluding inner join
```

##### **Compound assignement :each operator also has its own corresponding compound assignement if you add equal sign to it**
**=| , =& , =- , =^**

```python
# union + compound assignement
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}
my_set |= your_set
my_set
```

```python
# intersection + compound assignement
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}
my_set &= your_set
my_set
```

```python
# difference operator + compound assignement
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}
my_set -= your_set
my_set
```

```python
# symetric difference  + compound assignement
my_set = {1, 2, 3, 4, 5}
your_set = {2, 3, 4, 6}
my_set ^= your_set
my_set
```

# **Exception handling** :

### exception handling is core part of writing a robust and fault tolerant programs , to *anticipate* , *catch* and *respond* to errors in a structered way .

### **Exception handling is the process of catching and managing errors that occur during the execution of program , so your code doesn't crash unexpectedly**

### So Python provides **try , except , else , finally** *blocks* to gracefully handle errors .

## Example  of **try** , **except** : 
**try**: The block of code where you anticipate an error might occur.

**except**: This block runs if an error of the specified type is raised inside the try.

In this case, dividing by zero raises a **ZeroDivisionError**, which is then caught and handled.

```python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("You can't divide by zero!")
```

```python
try:
    x = 10 / y
except ZeroDivisionError:
    print("You can't divide by zero!")
except NameError:
    print("you can't do an operation on a variable that does''nt exist")
```

## Example  of **try** , **except** , **else** , **finally**  : 
**try**: The block of code where you anticipate an error might occur.

**except**: This block runs *if an error* of the specified type *is raised* inside the try.

**else** : is executed if there is *no exception* raised within code block.

**finally** : *Runs no matter what* - whether or not an exception occured . Useful for clean-up tasks like closing files or releasing ressources .

```python
try:
    x = 10 / 4
except ZeroDivisionError:
    print("You can't divide by zero!")
else:
    print('Division successful:', x)
finally:
    print('This block always runs.')
```

## Example  of **try** , **except** , **else** , **finally**  with *mutiple* exception :
### by using separate errors you can make errors more specific and useful 


```python
#converting a non number type to an integer raises a valueError
try:
    number = int('abc')
    result = 10 / number
except ValueError:
    print('That was not a valid number.')
except ZeroDivisionError:
    print("Can't divide by zero.")
```

```python
try:
    x = 1 / 0
except ZeroDivisionError as e:
    print(f'Error occurred: {e}') 
# using e lets you access the actual error 
# message or object for logging or debugging
```

# **you can raise built-in exceptions or create custom error messages**

```python
def check_age(age):
    if age < 0:
        raise ValueError('Age cannot be negative')
    return age

try:
    check_age(-5)
except ValueError as e:
    print(f'Error: {e}') # Error: Age cannot be negative
```

```python
def check_age(age):
    if age < 0:
        raise TypeError('Age cannot be negative')
    return age

try:
    check_age(-5)
except (TypeError,ValueError) as e:
    print(f'Error: {e}') # Error: Age cannot be negative
```

# Raise can be used to re raise the current exception . when called without arguements . it allows you to log or perform clean while still propagating the error up the call stack 

```python
def process_data(data):
    try:
        result = int(data)
        return result * 2
    except ValueError:
        print('Logging: Invalid data received')
        raise  # Re-raises the same ValueError

try:
    print(process_data('test'))
except ValueError:
    print('Handled at higher level')
```

# You can even create your own custom exception classes 

```python
class InsufficientFundsError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f'Insufficient funds: ${balance} available, ${amount} requested')

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(balance, amount)
    return balance - amount

try:
    new_balance = withdraw(100, 150)
except InsufficientFundsError as e:
    print(f'Transaction failed: {e}')
```

# chain exception using **from** keyword , showing the relationship between errors. 

```python
def parse_config(filename):
    try:
        with open(filename, 'r') as file:
            data = file.read()
            return int(data)
    except FileNotFoundError:
        raise ValueError('Configuration file is missing') from None
    except ValueError as e:
        raise ValueError('Invalid configuration format') from e


config = parse_config('config.txt')
```

### Use assert with  AssertionError :
#### Raise exceptions conditionally using **assert** statements , which are essentially shorthand for **raise** with **AssertionError**

```python
def calculate_square_root(number):
    assert number > 0, 'Cannot calculate square root of negative number'
    assert number == 0, 'Cannot calculate square root of null number'

    return number ** 0.5

try:
    result = calculate_square_root(0)
except AssertionError as e:
    print(f'Assertion failed: {e}')
```

```python
def calculate_square_root(number):
    if number > 0 :
        raise ValueError('Cannot calculate square root of negative number')
    if number == 0 :
        raise ValueError('Cannot calculate square root of null number')

    return number ** 0.5

try:
    result = calculate_square_root(0)
except Exception as e:
    print(f'Assertion failed: {e}')
```

```python
import datetime

class Email:
    def __init__(self, sender, receiver, subject, body):
        self.sender = sender
        self.receiver = receiver
        self.subject = subject
        self.body = body
        self.timestamp = datetime.datetime.now()
        self.read = False

    def mark_as_read(self):
        self.read = True

    def display_full_email(self):
        self.mark_as_read()
        print('\n--- Email ---')
        print(f'From: {self.sender.name}')
        print(f'To: {self.receiver.name}')
        print(f'Subject: {self.subject}')
        print(f"Received: {self.timestamp.strftime('%Y-%m-%d %H:%M')}")
        print(f'Body: {self.body}')
        print('------------\n')

    def __str__(self):
        status = 'Read' if self.read else 'Unread'
        return f"[{status}] From: {self.sender.name} | Subject: {self.subject} | Time: {self.timestamp.strftime('%Y-%m-%d %H:%M')}"
class Inbox:
    def __init__(self):
        self.emails = []

    def receive_email(self, email):
        self.emails.append(email)

    def list_emails(self):
        if not self.emails:
            print('Your inbox is empty.\n')
            return
        print('\nYour Emails:')
        for i, email in enumerate(self.emails, start=1):
            print(f'{i}. {email}')


    def read_email(self, index):
        if not self.emails:
            print('Inbox is empty.\n')
            return
        actual_index = index - 1
        if actual_index < 0 or actual_index >= len(self.emails):
            print('Invalid email number.\n')
            return
        self.emails[actual_index].display_full_email()

    def delete_email(self, index):
        if not self.emails:
            print('Inbox is empty.\n')
            return
        actual_index = index - 1
        if actual_index < 0 or actual_index >= len(self.emails):
            print('Invalid email number.\n')
            return
        del self.emails[actual_index]
        print('Email deleted.\n')
        
class User:
    def __init__(self, name):
        self.name = name
        self.inbox = Inbox()

    def send_email(self, receiver, subject, body):
        email = Email(sender=self, receiver=receiver, subject=subject, body=body)
        receiver.inbox.receive_email(email)
        print(f'Email sent from {self.name} to {receiver.name}!\n')

    def check_inbox(self):
        print(f"\n{self.name}'s Inbox:")
        self.inbox.list_emails()

    def read_email(self, index):
        self.inbox.read_email(index)

    def delete_email(self, index):
        self.inbox.delete_email(index)

def main():
    tory = User('Tory')
    ramy = User('Ramy')        
    
    tory.send_email(ramy, 'Hello', 'Hi Ramy, just saying hello!')
    ramy.send_email(tory, 'Re: Hello', 'Hi Tory, hope you are fine.')

    ramy.check_inbox()
    ramy.read_email(1)
    ramy.check_inbox()
    ramy.delete_email(1)
    ramy.check_inbox()

    
    
if __name__ == '__main__':
    main()
```
