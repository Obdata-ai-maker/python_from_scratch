Example : 

Let's say you want to get the number of pages in book objects created with the class below, or compare them and get a readable string of the objects. Here's what happens without special methods:

```python
class Book:
   def __init__(self, title, pages):
       self.title = title
       self.pages = pages

book1 = Book("Built Wealth Like a Boss", 420)
book2 = Book("Be Your Own Start", 420)

# print(len(book1)) # TypeError: object of type 'Book' has no len()
# print(str(book1)) # <__main__.Book object at 0x102ed2900>
print(book1 == book2) # False even though they have
```

1. len(book1) failed because Python doesn't know how to get the lenght of your book object without __len__()
2. str(book1) printed  <__main__.Book object at 0x102ed2900> when you don't use __str__() 
3. book1 == book2 resulted in false because Python just checks if both objects are the same in memory , not by content . 

```python
class Book:
    def __init__(self, title, pages):
       self.title = title
       self.pages = pages

    def __len__(self):
       return self.pages
    
    def __str__(self):
        return f"{self.title} has {self.pages}"
    
    def __eq__(self,other):
        return self.pages == other.pages

book1 = Book("Built Wealth Like a Boss", 420)
book2 = Book("Be Your Own Start", 420)

print(len(book1)) # 420
print(str(book1)) # Built Wealth Like a Boss has 420
print(book1 == book2) # True
```

### Another example is a shopping cart where you do : 
1. add items 
2. remove items 
3. get the number of items 
4. check what items are in the cart 
5. check if a specific item in the cart 
Return or display an item at a specific index in the cart . while you might have a method for adding items to a cart and removes certain items from  the cart , you can create a special method for all the other functionality .

__len__() to get the length of the items in the cart

__iter__() to loop through the items in the cart so you can see them

__contains__() to check if a specific item is in the cart

__getitem__() to return or display an item at a specific index in the cart


```python
class Cart:
   def __init__(self):
       self.items = []

   def add(self, item):
       self.items.append(item)

   def remove(self, item):
       if item in self.items:
           self.items.remove(item)
       else:
           print(f'{item} is not in cart')

   def list_items(self):
       return self.items

   def __len__(self):
       return len(self.items)

   def __getitem__(self, index):
       return self.items[index]

   def __contains__(self, item):
       return item in self.items

   def __iter__(self):
       return iter(self.items)
```

```python
cart = Cart()
cart.add('Laptop')
cart.add('Wireless mouse')
cart.add('Ergo keyboard')
cart.add('Monitor')

for item in cart:
   print(item, end=' ') # Laptop Wireless mouse Ergo keyboard Monitor


print(len(cart)) # 4
print(cart[3]) # Monitor ---> is equivalent to __getitem__
```

```python
cart.remove('Ergo keyboard')

print(cart.list_items()) # ['Laptop', 'Wireless mouse', 'Monitor']

cart.remove('banana') # banana is not in cart
```

## **How to Handle Object Attributes Dynamically?**

#### attributes are are variables that belong to an object . that means they hold they hold data of that describes the state or behaviour of the object .
#### but sometimes you might not know wich attributes you need until your programm is running . Imagine you are writing a script that receives attributes names from configuration files . those are not attributes you can hardcode ahead of time .

#### so you can access , modify , add or even delete attributes using : 
**getattr()** , **setattr()** , **hasattr()** , **delattr()**

1. #### **getattr()** : lets you read an attribute object when you don't know its name until runtime . if attribute doesn't exist it raises an **AttributeError** unless you provide a default value . 
#### getattr(object, attribute_name, default_value) 
   

```python
class Person: 
    def __init__(self, name, age): 
        self.name = name 
        self.age = age 

person = Person('John Doe', 30) 
 
print(getattr(person, 'name')) # John Doe 
print(getattr(person, 'age')) # 30 
#print(getattr(person, 'city')) # AttributeError 
print(getattr(person, 'city', 'value not found')) # Milano
```

## the real power of dynamic attributes is when they come from variables , user input or some file . in that case you can't use object.attribute_name because the attribute name is not fixed

```python
class Person: 
    def __init__(self, name, age): 
        self.name = name 
        self.age = age 

person = Person('John Doe', 30)

attr_name = input('Enter the attribute you want to see: ')
print(getattr(person, attr_name, 'Attribute not found'))

# so if you type age or name you get the data of object , 
# but if else you get Attribute not found
```

### **dir**() lets list all methods and attributes of an object

#### **callable**() returns **True** is the object passed to it can be called like a function or method 

```python
class Person: 
    def __init__(self, name, age): 
        self.name = name 
        self.age = age 

person = Person('John Doe', 30)

# Loop through all attributes of the person object with dir() function
for attr in dir(person):
    # Ignore dunder methods like __init__ or __str__ and regular methods
    if not attr.startswith('__') and not callable(getattr(person, attr)): 
        value = getattr(person, attr)
        print(f'{attr}: {value}')

# Output
# age: 30
# name: John Doe
```

1. #### **setattr()** : lets you create or update an attribute object when you don't know its name until runtime .
#### **setattr**(object, attribute_name, value) 
##### example with configuration variable / file

```python
class Configuration:
    pass

# Data loaded at runtime (like from a config or env file)
settings_data = {
    'server_url': 'https://api.example.com',
    'timeout_sec': 30,
    'max_retries': 5
}

config_obj = Configuration()

# Dynamically set attributes using dictionary keys and values
for attr_name, attr_value in settings_data.items():
    setattr(config_obj, attr_name, attr_value)

print(config_obj.server_url) # https://api.example.com
print(config_obj.timeout_sec) # 30
```

1. #### **hasattr()** : lets you check if an object has this attribute  .
#### **hasattr**(object, attribute_name) and return **True/False**

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

product_a = Product('T-Shirt', 25)

required_attributes = ['name', 'price', 'inventory_id']

for attr in required_attributes:
    if not hasattr(product_a, attr):
        print(f"ERROR: Product is missing the required attribute: '{attr}'")
    else:
        # Access the attributes dynamically once their existence is confirmed
        print(f'{attr}: {getattr(product_a, attr)}')

# Output:
# name: T-Shirt
# price: 25
# ERROR: Product is missing the required attribute: 'inventory_id'
```

1. #### **delattr()** : lets you check if an object has this attribute  .
#### **delattr**(object, attribute_name)

#### for example imagine an object has been fully processed , then you decide to clean up any sensitive or temporary attributes that might exist before saving the final versiion , after that you use **dir()** to loop through the remainig attributes .

```python
class UserSession:
    def __init__(self, user_id, token):
        self.user_id = user_id
        self.auth_token = token # sensitive
        self.temp_counter = 0 # temporary

session = UserSession(101, 'a1b2c3d4e5')

# List of attributes to remove dynamically before "saving" the session
attributes_to_clean = ['auth_token', 'temp_counter']

# Dynamically remove specified attributes
for attr in attributes_to_clean:
    if hasattr(session, attr):
        delattr(session, attr)
        print(f'Removed attribute: {attr}')

print('\nFinal attributes remaining:')

# Loop through the remaining attributes with dir()
for attr in dir(session):
    # Ignore dunder methods like __init__ or __str__ and regular methods
    if not attr.startswith('__') and not callable(getattr(session, attr)):
        print(f' - {attr}: {getattr(session, attr)}')

# Output:
# Removed attribute: auth_token
# Removed attribute: temp_counter

# Final attributes remaining:
#  - user_id: 101
```

# OOP and encapsulation

## encapsulation definition 
Encapsulation is the principle of bundling an object's data and behavior together while controlling access to its internal state through a well-defined public interface.

From that principle, several benefits follow.

Limit direct access to the object's state.

    Other objects cannot arbitrarily modify its attributes.

    The object remains responsible for maintaining its own consistency.

Define how the object may be used.

    Instead of exposing internal data, you expose operations (methods).These methods form the object's public interface or API.

Enforce business and technical rules.

    Business: "A bank account cannot have a negative balance."

    Technical: "This file must be closed after writing."

Hide implementation details.

    Other parts of the program don't need to know how the object works, only what it can do.

Reduce coupling.

    Because other code depends only on the interface, you can change the internal implementation without breaking the rest of the system.

## from architecture stand point 
encapsulation means basically protect object's internal state, by limiting access to the attributes from tampering , and defining the interaction with outside code by using an interface . to keep data integrity , hide complexity and even allows to enforce technical and buisness rules of the object  the code can updated without disturbing knowing that outside code will touch only the interface that you expose .

```python
class Wallet:
   def __init__(self, balance):
       self.__balance = balance

   def deposit(self, amount):
       if amount > 0:
           self.__balance += amount

   def withdraw(self, amount):
       if 0 < amount <= self.__balance:
           self.__balance -= amount
  
   def get_balance(self):
       return self.__balance


acct_one = Wallet(100)
acct_one.deposit(50)
print(acct_one.get_balance()) # 150

acct_two = Wallet(450)
acct_two.withdraw(28)
print(acct_two.get_balance()) # 422

acct_two.deposit(150)
print(acct_two.get_balance()) # 572
```

# private attributes and private methods

```python
class Wallet:
   def __init__(self):
       self.__balance = 0

   def __validate(self, amount):
       if amount < 0:
           raise ValueError('Amount must be positive')

   def deposit(self, amount):
       self.__validate(amount)
       self.__balance += amount

   def withdraw(self, amount):
       self.__validate(amount)
       if amount > self.__balance:
           raise ValueError('Insufficient funds')
       self.__balance -= amount

   def get_balance(self):
       return self.__balance

acct_one = Wallet()
acct_one.deposit(3)
print(acct_one.get_balance()) # 3

acct_one.deposit(50)
print(acct_one.get_balance()) # 53

# acct_one.deposit(-4)  # ValueError: Amount must be positive
# acct_one.withdraw(-8) # ValueError: Amount must be positive
acct_one.withdraw(58) # ValueError: Insufficient funds
```

### to make a setter to create the radius for example : you have to define another method with the same name and use @<property_name>.setter above it : 
 so uisng self.radius inside ___init___ ensure that the setter runs during object creation , so invalid radius values are caught immediatly 

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius # Calling the setter

    @property
    def radius(self):  # A getter to get the radius
        return self._radius

    @radius.setter
    def radius(self, value):  # A setter to set the radius
        if value <= 0:
            raise ValueError('Radius must be positive')
        self._radius = value
```

# just how you can set and get , you can create a custom logic to delete ( create a logic statement to delete )

```python
class Circle: 
    def __init__(self,radius):
         self.radius = radius # calling the setter 
    
    @property
    def radius(self):
         return self._radius  # a getter 
    # if we use self.radius it will be recursion 

    @radius.setter
    def radius(self,value): # setter to set the radius 
          if value <= 0 : 
               raise ValueError("Radius must be positive")
          self._radius = value
    
    @radius.deleter
    def radius(self): # deleter to delete the radius 
        print("Deleting radius")
        del  self._radius
```

```python
# Create circle object with a radius
my_circle = Circle(33)
print("Initial radius:", my_circle.radius)  # 33

# Delete the radius
# This calls the deleter
del my_circle.radius # Deleting radius...
print("Radius deleted!") # Radius deleted!


# Try to access radius after deletion
try:
    print(my_circle.radius)
except AttributeError as e:
    print("Error:", e) # Error: 'Circle' object has no attribute '_radius'
```

# Encapsulation Lab

```python
class Employee:
    _base_salaries = {
        'trainee': 1000,
        'junior': 2000,
        'mid-level': 3000,
        'senior': 4000,
    }

    def __init__(self, name, level):
        self.name = name
        self.level = level
        self.salary = Employee._base_salaries[level]

    def __str__(self):
        return f'{self.name}: {self.level}'

    def __repr__(self):
        return f"Employee('{self.name}', '{self.level}')"

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, new_name):
        if not isinstance(new_name, str):
            raise TypeError("'name' must be a string.")
        self._name = new_name
        print(f"'name' updated to '{self.name}'.")

    @property
    def level(self):
        return self._level

    @level.setter
    def level(self, new_level):
        if not isinstance(new_level, str):
            raise TypeError("'level' must be a string.")
        if new_level not in Employee._base_salaries:
            raise ValueError(f"Invalid value '{new_level}' for 'level' attribute.")
        if hasattr(self, '_level') and new_level == self.level:
            raise ValueError(f"'{self.level}' is already the selected level.")
        if hasattr(self, '_level') and Employee._base_salaries[new_level] < Employee._base_salaries[self.level]:
            raise ValueError("Cannot change to lower level.")
        print(f"'{self.name}' promoted to '{new_level}'.")
        self.salary = Employee._base_salaries[new_level]
        self._level = new_level

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, new_salary):
        if not isinstance(new_salary, (int, float)):
            raise TypeError("'salary' must be a number.")
        if hasattr(self, '_level') and new_salary < Employee._base_salaries[self.level]:
            raise ValueError(f'Salary must be higher than minimum salary ${Employee._base_salaries[self.level]}.')
        self._salary = new_salary
        print(f'Salary updated to ${self.salary}.')

charlie_brown = Employee('Charlie Brown', 'trainee')
print(charlie_brown)
print(f'Base salary: ${charlie_brown.salary}')
charlie_brown.level = 'junior'
```

# Inheritance 
is a concept that a child class can inherit all parent class and at  the same extends or overrides where necessary 

```python
class Animal :
    def __init__(self,name):
        self.name = name
    
    def sound(self):
        return f'{self.name} makes a sound'
    
class Dog(Animal): # here the parent class is Animal and child class is Dog
    # in this case Dog will inherit all parent class attributes and methods
    bark = "wooof wooof "

jack = Dog("ringo")
print(jack.sound())
print(jack.bark)
```

# Override method

```python
class Animal :
    def __init__(self,name):
        self.name = name
    
    def sound(self):
        return f'{self.name} makes a sound'
    
class Dog(Animal): # here the parent class is Animal and child class is Dog
    # in this case Dog will inherit all parent class attributes and methods
    bark = "wooof wooof "
    def sound(self):
        return f'{self.name} barks {self.bark}'
        

jack = Dog("ringo")
print(jack.sound())
print(jack.bark)
```

# Extend method
if we want keep the return value of sound() and add bark class variable later you can extend sound() by using super() function 

```python
class Animal :
    def __init__(self,name):
        self.name = name
    
    def sound(self):
        return f'{self.name} makes a sound'
    
class Dog(Animal): # here the parent class is Animal and child class is Dog
    # in this case Dog will inherit all parent class attributes and methods
    bark = "wooof wooof "
    def sound(self):
        base = super().sound()
        return f'{base} , then {self.name} barks {self.bark}'
        

jack = Dog("ringo")
print(jack.sound())
print(jack.bark)
```

# Multiple Inheritance
Child class can inherit from multiple parent classes 

```python
class Parent:
    # Attributes and methods for Parent
    pass

class Child:
    # Attributes and methods for Child
    pass

class GrandChild(Parent, Child):
    # GrandChild inherits from both Parent and Child
    # GrandChild can combine or override behavior from each
    pass
```

a way to demonstrate multiple inheritances is with a frog , wich can both walk on land and swim on water :

```python
class Walker:
    # Attributes and methods for Parent
    def walk(self):
        return 'i can walk on land'

class Swimmer:
    # Attributes and methods for Child
    def swim(self):
        return 'i can swim in water'

class Amphibian(Walker,Swimmer):
    def __init__(self,name):
        self.name = name
    
    def introduce(self):
        return f"i'm {self.name} the frog . {self.walk()} and {self.swim()}"

frog = Amphibian('Freddy')
print(frog.introduce())
# Output: 
```
